```python
df = pd.read_csv("people-100.csv",index_col = "Name")
```

```python
df = df.fillna({"Type2":"None"})
print(df.to_string())
```

```python
df["Type1"] = df["Type1"].replace({"Grass" : "GRASS"})
print(df.to_string())
```

```python
df["Legendary"] = df["Legendary"].astype(bool)
print(df)
```

```python
df = df.drop_duplicates()
print(df.to_string())
```
