```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
plt.style.available
plt.style.use("dark_background")
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y,color = "blue");
```

```python
x = np.linspace(0, 10, 100)
plt.plot(x,np.sin(x),linestyle = "dashed",color = "red",label = "sin(x)");
plt.plot(x,np.cos(x),linestyle = "dotted",color = "green",label = "cos(x)");
plt.xlabel("X")
plt.ylabel("sin(x)")
plt.legend()
plt.title("sinx and cosx")

```

```python
x = [1,2,3,4]
y = [10,20,30,40]
fig,ax = plt.subplots(figsize= (5,5))
ax.plot(x,y);
ax.set(title = "simple plot",xlabel = "x-axis",ylabel = "y-axis")
```

```python
# OO API
rng = np.random.RandomState(0)
x = rng.randn(100)
y = rng.randn(100)
fig,ax = plt.subplots()
color = rng.rand(100)
sizes = 1000*np.random.rand(100)
img1 = ax.scatter(x,y,c=color,s=sizes,alpha = 0.3,cmap = "viridis");
fig.colorbar(img1)
```

```python
soft_drink_prices = {"Coca-Cola": 2.50, 
                     "Pepsi": 2.50, 
                     "Sprite": 2.00, 
                     "Fanta": 2.00, 
                     "Mountain Dew": 2.50}
fig,ax = plt.subplots()
ax.bar(soft_drink_prices.keys(),soft_drink_prices.values(),color = "orange")
ax.set_title("Soft Drink Prices")
ax.set_xlabel("Soft Drink")
ax.set_ylabel("Price ($)")
```

```python
student_height = np.random.normal(170,10,1000)
plt.hist(student_height,bins = 30,color = "blue",edgecolor = "black");
```

```python
fig , ((ax1 , ax2) , (ax3,ax4)) = plt.subplots(nrows = 2, ncols = 2 , figsize = (10,5))
ax1.plot(x,x/2)
ax1.plot(x,x*2)
ax2.scatter(np.random.random(10),np.random.random(10))
ax3.bar(soft_drink_prices.keys(),soft_drink_prices.values(),color = "orange")
ax4.hist(student_height,bins = 30,color = "blue",edgecolor = "black");
```

```python
fig,ax = plt.subplots(nrows = 2 , ncols = 2 , figsize = (10,5))
ax[0][0].plot(x,x/2)
ax[0][0].plot(x,x*2)
ax[0][1].scatter(np.random.random(10),np.random.random(10))
ax[1][0].bar(soft_drink_prices.keys(),soft_drink_prices.values(),color = "orange")
ax[1][1].hist(student_height,bins = 30,color = "blue",edgecolor = "black");
```
