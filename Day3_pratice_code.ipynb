```python
import numpy as np
```

```python
x = np.array([1,2,3,4],dtype =float)
y = np.array([3,5,7,9],dtype = float)
n = len(x)
w,b = 0.0,0.0
```

```python
def predict (x,w,b):
    y_pred = x*w+b
    return y_pred
```

```python
def compute_mse(y_pred,y_true):
    return np.mean((y_true-y_pred)**2)
```

```python
y_pred = predict(x,w,b)
mse = compute_mse(y_pred,y)
errors = y_pred-y
```

```python
dw = (2/n) * np.sum(errors * x)
db = (2/n) * np.sum(errors)
ln = 0.01
w = w - ln*dw
b = b - ln*db
print("MSE ban dau: ",mse)
print("dw: ", dw)
print("db: ",db)
print("w moi: ",w)
print("b moi: ",b)
```
