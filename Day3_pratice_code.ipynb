```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4],dtype =float)
y = np.array([3,5,7,9],dtype = float)
n = len(x)
w,b = 0.0,0.0
```

```python
def predict (x,w,b):
    y_pred = x*w+b
    return y_pred
```

```python
def compute_mse(y_pred,y_true):
    return np.mean((y_true-y_pred)**2)
```

```python
def compute_gradient(x,y,y_pred):
    n = len(x)
    error = y_pred - y
    dw = (2/n) * np.sum((error*x))
    db = (2/n) * np.sum((error))
    return dw,db
```

```python
loss_history = []
epochs = 1000
```

```python
for epoch in range (epochs):
    y_pred = predict(x,w,b)
    errors = y_pred - y
    mse = compute_mse(y_pred,y)
    loss_history.append(mse)
    dw,db = compute_gradient(x,y,y_pred)
    ln = 0.01
    w = w - ln*dw
    b = b - ln*db
```

```python
print("w : ",w)
print("b : ",b)
print("mse : " , compute_mse(predict(x,w,b),y))
```

```python
y_pred = predict(x,w,b)
plt.scatter(x,y,label = "actual data")
plt.plot(x,y_pred,label = "prediction")
plt.xlabel("study hours")
plt.ylabel("score")
plt.legend()
```

```python
plt.plot(loss_history)
plt.xlabel("epoch")
plt.ylabel("mse")
```

```python
mse = compute_mse(y_pred,y)
mae = np.mean(np.abs(y_pred-y))
rmse = np.sqrt(mse)
ss_res = np.sum((y-y_pred)**2)
ss_tot = np.sum((y-np.mean(y))**2)
r2 = 1 - ss_res/ss_tot
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2:", r2)


```
