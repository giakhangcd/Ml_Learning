```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./california_cities.csv")
df["area_total_km2"].sort_values(ascending = False) 
```

```python
lat,lon = df['latd'],df['longd'] 
population,area = df['population_total'],df['area_total_km2']
plt.figure(figsize=(8,6))
plt.scatter(lon,lat, c = np.log10(population) ,s = area, cmap = "viridis",linewidth = 0.5,alpha = 0.5)
plt.axis('equal')
plt.xlabel("Longitude")
plt.ylabel("Latitude")  
plt.colorbar(label = 'log$_{10}$(population)')
plt.clim(3,7)
plt.title('california_cities population')
area_range = [50,100,300,500]
for area in area_range:
    plt.scatter([],[],s = area , label = str(area) + 'km$^2$' )
    plt.legend() 
```
